```python
import cv2
import numpy as np

# === Load Video and Extract Images into frames array ===
def extract_frames(video_path):
    # Open the video file
    cap = cv2.VideoCapture(video_path)
    frames = []
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        frames.append(gray_frame)
    cap.release()
    return frames

# Load Video, Extract Images into frames array, cropping if needed
# so that spatial dimensions are integer multiples of 8x8
def crop_frames_to_multiples_of_8(video_path):
    # Open the video file
    cap = cv2.VideoCapture(video_path)
    frames = []

    while cap.isOpened():
        ret, image = cap.read()
        if not ret:
            break

        # Get the original dimensions of the frame
        height, width = image.shape[:2]

        #print(f" dimensions h: {height} ; w: {width}" )
        
        # Calculate new dimensions that are multiples of 8
        new_height = (height // 8) * 8
        new_width = (width // 8) * 8
        
        #print(f"new dimensions h: {new_height} ; w: {new_width}" )

        # Crop the frame to these dimensions
        frame = image[:new_height, :new_width]

        # Append the Y cropped frame to the list
        gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        frames.append(gray_frame)

    cap.release()
    return frames
```
